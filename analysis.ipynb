```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import seaborn as sns
```

## Workflow

1. Data import
   - Data is RPKM normalized
2. Data wrangling
3. Determining Intra- and Intergroup Sample Variability and Outliers
   - PCA
   - Pearson’s coefficient
   - Spearman’s rank correlation coefficient


```python
df = pd.read_csv("./sample_data/GSE116583_transplant.am.htseq.all.rpkm.txt", sep="\t")
info = pd.read_csv("./sample_data/GSE116583_series_matrix.txt", sep="\t", skiprows=26)
df.head()
```

```python
info.head()
```

```python
# Programmatically ensure data is in same order as metadata
dfcols = list(df.columns[1:])
samplerow = (
    info.loc[info["!Sample_title"] == "!Sample_description"].iloc[0].to_list()[1:]
)

assert all(n in m for n, m in zip(df.columns[1:], samplerow))

print(dfcols)
print(samplerow)
```

```python
# Rename df columns
cols = [
    "Naive_01",
    "Naive_02",
    "Naive_03",
    "Naive_04",
    "Transplant_2h_01",
    "Transplant_2h_02",
    "Transplant_2h_03",
    "Transplant_2h_04",
    "Transplant_24h_05",
    "Transplant_24h_06",
    "Transplant_24h_07",
    "Transplant_24h_08",
]

df = df.set_axis(
    ["Gene"] + cols,
    axis=1,
)

dfcounts = df.copy().drop(columns=["Gene"])  # Drop Gene column

# Log-transform data and transpose dataframe for PCA
dfnorm = dfcounts.apply(lambda x: np.log10(x + 1))
# Scale data (this scaler is probably different from the one used in the paper)
scaler = StandardScaler()
data_scaled = scaler.fit_transform(dfnorm)
dfscaled = pd.DataFrame(data_scaled, columns=dfnorm.columns)
df_T = dfscaled.T
```

## PCA


```python
# Run PCA and store results in df_pca
pca = PCA()
pca_results = pca.fit_transform(df_T)

explained_variance = pca.explained_variance_ratio_

groups = df_T.index.str.replace(r"_\d+$", "", regex=True)
df_pca = pd.DataFrame(
    {
        "PC1": pca_results[:, 0],
        "PC2": pca_results[:, 1],
        "Group": groups,
        "Sample": df_T.index,
    }
)
```

```python
# Scree Plot
plt.figure(figsize=(6, 5))

# Create a scree plot to visualize the explained variance
plt.plot(
    range(1, len(explained_variance) + 1),  # x-axis
    explained_variance * 100,  # convert explained variance in percentage
    marker="o",  # add a marker at each value
)
plt.title("Scree Plot of Explained Variance for Principal Components")
plt.xlabel("Principal Component")
plt.ylabel("Explained Variance (in %)")
```

```python
# PCA Scatter Plot
plt.figure(figsize=(5, 5))
for group in df_pca["Group"].unique():
    subset = df_pca[df_pca["Group"] == group]
    plt.scatter(subset["PC1"], subset["PC2"], label=group, s=100)

plt.xlabel(f"PC1 ({explained_variance[0] * 100:.1f}% var)")
plt.ylabel(f"PC2 ({explained_variance[1] * 100:.1f}% var)")
plt.legend()
```

```python
# Pearson’s correlation matrix. (Non-log-normalized data)
matrix = dfcounts.corr(method="pearson")

plt.figure(figsize=(6, 5))
sns.heatmap(matrix, cmap="Blues", linewidths=0.5)
plt.title("Pearson Correlation Heatmap")
plt.show()
```

```python
# Number of genes at a given RPKM value for each sample (bins = 120, bin size = 0.1)
# For example, Native_01 has 43430 values greater than 0.0, 15620 > 0.1 ... up to 11.9 (= 120 bins)

# Define bins and output df
rpkm_bins = np.arange(0, 12.1, 0.1)
rpkm_hist = pd.DataFrame(index=rpkm_bins[:-1])

# Compute counts of genes > cutoff
for col in dfcounts.columns:
    vals = dfcounts[col].values
    # For each cutoff, count how many genes >= cutoff
    rpkm_hist[col] = [(vals >= cutoff).sum() for cutoff in rpkm_bins[:-1]]


fig, ax = plt.subplots(figsize=(10, 6))
# Main plot
sns.lineplot(data=rpkm_hist, ax=ax)
ax.set_xlim([0, 10])
ax.set_ylim([0, 20000])
ax.set_xlabel("RPKM")
ax.set_ylabel("Number of genes > RPKM cutoff")
ax.set_title("Number of genes at a given RPKM for each sample")

# Define inset limits
x1, x2, y1, y2 = 0, 2, 10000, 20000

# Create inset axes inside main axes
axins = ax.inset_axes(
    [0.3, 0.55, 0.4, 0.4]
)  # [x0, y0, width, height] in relative coords
sns.lineplot(data=rpkm_hist, ax=axins, legend=False)

# Set zoom area in inset
axins.set_xlim(x1, x2)
axins.set_ylim(y1, y2)

# Indicate inset zoom region
ax.indicate_inset_zoom(axins, edgecolor="black")
```

The samples start to align at RPKM 1, so the low counts threshold can be set at 1 RPKM.


```python
# Scatterplots comparing the expression of individual genes between two samples
# Expression data is scaled by log2 and negative values are set to 0 as in the paper

# Most correlated samples within group 1
data = dfcounts[["Naive_01", "Naive_03"]].apply(np.log2).clip(0)
ax = sns.scatterplot(
    x="Naive_01", y="Naive_03", data=data, color="black", edgecolor="black"
)
ax.set(xlabel="Naive_01 (log2)", ylabel="Naive_03 (log2)")
plt.show()

# Least correlated samples within group 2
data = dfcounts[["Transplant_2h_01", "Transplant_2h_03"]].apply(np.log2).clip(0)
ax = sns.scatterplot(
    x="Transplant_2h_01",
    y="Transplant_2h_03",
    data=data,
    color="black",
    edgecolor="black",
)
ax.set(xlabel="Transplant_2h_01 (log2)", ylabel="Transplant_2h_03 (log2)")
plt.show()

# Least correlated samples within the data set
data = dfcounts[["Transplant_24h_06", "Naive_04"]].apply(np.log2).clip(0)
ax = sns.scatterplot(
    x="Transplant_24h_06",
    y="Naive_04",
    data=data,
    color="black",
    edgecolor="black",
)
ax.set(xlabel="Transplant_24h_06 (log2)", ylabel="Naive_04 (log2)")
plt.show()
```
